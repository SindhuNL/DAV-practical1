```python
import pandas as pd 
import numpy as np
```

```python
data = {
    'Study_Hours': [2, 4, 6, 8, 10, 3, 5, 7, 9, 12],
    'Attendance': [60, 65, 70, 75, 80, 62, 68, 78, 85, 95],
    'Assignment_Score': [40, 50, 55, 65, 70, 45, 60, 72, 80, 90],
    'Exam_Score': [45, 50, 60, 65, 75, 48, 58, 70, 82, 95]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Dataset:")
print(df)
print("\nStatistical Summary:")
print(df.describe())
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
normalized_data = scaler.fit_transform(df)
```

```python
normalized_df = pd.DataFrame(normalized_data, columns=df.columns)
```

```python
print("Normalized Dataset:")
print(normalized_df.round(2))
```

```python
print("\nMinimum values:")
print(normalized_df.min())
```

```python
print("\nMaximum values:")
print(normalized_df.max())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
standardized_data = scaler.fit_transform(df)
```

```python
standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)
```

```python
print("Standardized Dataset:")
```

```python
print(standardized_df.round(2))
```

```python
print("\nMean of each feature:")
print(standardized_df.mean().round(2))
```

```python
print("\nStandard deviation of each feature:")
print(standardized_df.std(ddof=0).round(2))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
```

```python
pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)
```

```python
pca_df = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)
```

```python
print("PCA Transformed Data:")
print(pca_df.round(2))
```

```python
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

```python
print("\nTotal Variance Retained:")
print(pca.explained_variance_ratio_.sum())
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    pca_df['PC1'],
    pca_df['PC2'],
    color='blue',
    s=60)
```

```python
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Dimensionality Reduction")
plt.grid(True)
plt.show()
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Select numerical features
X = df.select_dtypes(include=['number'])

# Standardize data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("PCA completed successfully!")
print(X_pca.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(X_pca[:, 0], X_pca[:, 1], color='blue')

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Dimensionality Reduction")

plt.grid(True)
plt.show()
```
